```python
import numpy as np

# Convert text to binary representation
def text_to_binary(text):
    binary_message = ''.join(format(ord(char), '08b') for char in text)  # Convert each char to 8-bit binary
    return binary_message

# Convert binary string into an array of integers
def binary_to_array(binary_message, n):
    # Convert the binary string to an array of integers
    binary_array = np.array([int(bit) for bit in binary_message], dtype=np.int32)
    
    # Pad with zeros or truncate to fit the size `n+1`
    if len(binary_array) < n+1:
        binary_array = np.concatenate([binary_array, np.zeros(n+1 - len(binary_array), dtype=int)])
    else:
        binary_array = binary_array[:n+1]
    
    return binary_array

```

```python
def binary_to_text(binary_array):
    # Join the binary array into a single binary string
    binary_string = ''.join(str(bit) for bit in binary_array)
    
    # Split the binary string into 8-bit chunks (since each character is 8 bits)
    text = ''.join(chr(int(binary_string[i:i+8], 2)) for i in range(0, len(binary_string), 8))
    return text
```
