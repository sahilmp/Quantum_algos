```python
%%time
import pandas as pd
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.backends import default_backend

def hash_cell(value):
    chf = hashes.Hash(hashes.SHA256(), backend=default_backend())
    chf.update(str(value).encode('utf-8'))  # Convert the value to a string and encode it
    return chf.finalize().hex()

def verify_hashed_dataset(original_df, hashed_df):
    for col in original_df.columns:
        for idx, value in original_df[col].items():
            # Recalculate the hash of the original cell value
            recalculated_hash = hash_cell(value)
            
            # Compare with the corresponding hashed cell
            if hashed_df.loc[idx, col] != recalculated_hash:
                return False  # Mismatch found
    return True  # All cells match

# Load your dataset
df = pd.read_csv("creditcard.csv")  # Replace with your actual file path

# Hash each cell in the dataset
hashed_df = df.applymap(hash_cell)

# Verify the hashed dataset
is_verified = verify_hashed_dataset(df, hashed_df)

# Display the results
print("Original DataFrame:")
print(df.head())
print("\nHashed DataFrame:")
print(hashed_df.head())
print(f"\nVerification Result: {'Successful' if is_verified else 'Failed'}")

# Save the hashed dataset
hashed_df.to_csv("hashed_dataset.csv", index=False)
```
