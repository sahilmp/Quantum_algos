```python
%%time
import pandas as pd
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
import os
import base64

# Step 1: Generate RSA Keys (Public and Private)
def generate_keys():
    private_key = rsa.generate_private_key(
        public_exponent=65537,
        key_size=2048,
    )
    public_key = private_key.public_key()
    return private_key, public_key

# Step 2: Encrypt the entire column
def encrypt_column(public_key, column):
    # Convert column to a single string
    column_string = ','.join(map(str, column.values))
    
    # Generate a random AES key
    aes_key = os.urandom(32)  # 256-bit key
    
    # Encrypt the column using AES
    iv = os.urandom(16)  # Initialization vector
    cipher = Cipher(algorithms.AES(aes_key), modes.CFB(iv))
    encryptor = cipher.encryptor()
    ciphertext = encryptor.update(column_string.encode()) + encryptor.finalize()
    
    # Encrypt the AES key using RSA
    encrypted_key = public_key.encrypt(
        aes_key,
        padding.OAEP(
            mgf=padding.MGF1(algorithm=hashes.SHA256()),
            algorithm=hashes.SHA256(),
            label=None
        )
    )
    
    return (encrypted_key, iv, ciphertext)

# Step 3: Decrypt the entire column
def decrypt_column(private_key, encrypted_data):
    encrypted_key, iv, ciphertext = encrypted_data
    
    # Decrypt the AES key using RSA
    aes_key = private_key.decrypt(
        encrypted_key,
        padding.OAEP(
            mgf=padding.MGF1(algorithm=hashes.SHA256()),
            algorithm=hashes.SHA256(),
            label=None
        )
    )
    
    # Decrypt the column using AES
    cipher = Cipher(algorithms.AES(aes_key), modes.CFB(iv))
    decryptor = cipher.decryptor()
    decrypted_column_string = decryptor.update(ciphertext) + decryptor.finalize()
    
    # Convert the decrypted string back to a list
    return decrypted_column_string.decode().split(',')

# Main Code to Encrypt and Decrypt a Dataset
# Generate RSA Keys
private_key, public_key = generate_keys()

# Load Dataset
df = pd.read_csv("fraudTrain.csv")  # Replace with your dataset path

# Encrypt the dataset column by column
encrypted_columns = {
    column: encrypt_column(public_key, df[column]) for column in df.columns
}

# Create a DataFrame for the encrypted data
encrypted_df = pd.DataFrame(
    {
        column: [
            base64.b64encode(encrypted_columns[column][0]).decode(),  # Encrypted key
            base64.b64encode(encrypted_columns[column][1]).decode(),  # IV
            base64.b64encode(encrypted_columns[column][2]).decode(),  # Ciphertext
        ]
        for column in df.columns
    }
).transpose()
encrypted_df.columns = ["encrypted_key", "iv", "ciphertext"]

# Decrypt the dataset column by column
decrypted_columns = {
    column: decrypt_column(private_key, encrypted_columns[column]) for column in df.columns
}

# Convert decrypted columns back to a DataFrame
decrypted_df = pd.DataFrame(decrypted_columns)

# Verify results
print("Original DataFrame:")
print(df.head())
print("\nEncrypted DataFrame (Column-wise):")
print(encrypted_df)
print("\nDecrypted DataFrame:")
print(decrypted_df.head())
```
